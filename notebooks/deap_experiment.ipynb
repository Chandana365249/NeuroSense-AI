# NeuroSense AI: DEAP exploration

Exploratory notebook. The pipeline in `backend/app` is authoritative; this only calls it.
Run with the project `.venv` as kernel.

```python
import os, sys
sys.path.insert(0, os.path.abspath('../backend'))
os.chdir(os.path.abspath('../backend'))

from app.core.config import get_settings
from app.ml.factory import make_dataset, make_preprocessor, make_feature_extractor

settings = get_settings()
dataset = make_dataset(settings)
print(dataset.participant_ids())
```

```python
eeg, ratings = dataset.load_trial(dataset.participant_ids()[0], 0)
result = make_preprocessor(settings).preprocess(eeg, has_baseline=True)
features = make_feature_extractor(settings).extract(result.segments)
print('ratings (valence, arousal, dominance, liking):', ratings)
print('segments:', result.segments.shape, 'features:', features.shape)
```
